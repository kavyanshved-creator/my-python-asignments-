```python
"""1.
Write a Python script that takes your current Spotify listening time
in minutes and checks if it is above 120 minutes; if yes, print
'You are a true music fan!', otherwise print 'Keep listening!'."""

listening_time = int(input("Enter your current spotify listining time in minuites : "))

if listening_time > 120 : 
    print("You are a true musice fan! ")
else :
    print("Keep listioning! ")
    
```

```python
"""2.
Create a Python program that asks the user to enter their Zomato
order amount and checks if it is above 300; if yes, print 'Eligible
for free delivery', else print 'Delivery charges apply'."""

zomato_order_amount = int(input("enter your zomato amount : "))

if zomato_order_amount > 300 :
    print("Eligible for free delivery ")
else:
    print("Delivery charges may apply ")
```

```python
"""3.
Build a Python script that takes your Flipkart cart total and applies 
the following logic: if total > 2000, print 'You get a 10% discount'; 
elif total > 1000, print 'You get a 5% discount'; else print 'No discount 
available'."""

Flipkart_total = int(input("Enter the total amount "))

if Flipkart_total >= 2000 :
    print("You get a 10% discount")
elif Flipkart_total >= 1000 :
    print("You get a 5% discount")
else :
    print("No discount available ")
```

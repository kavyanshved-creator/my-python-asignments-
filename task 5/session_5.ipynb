```python
"""1.
Create a list called playlist_ids with 5 song IDs (as integers)
that you might see in a Spotify playlist, and print the list.
"""
playlist_ids = [101,202,303,404,505]
print(playlist_ids)
```

```python
"""2.
Add two more song IDs to your playlist_ids list using both append() 
and extend(), then print the updated list.<br><br><em><strong>Hint:</strong> 
Use append() for a single ID and extend() for adding multiple IDs at once.</em>
"""

playlist_ids = [101,202,303,404,505]
print(playlist_ids)

playlist_ids.append(203) # append add single value list
print(playlist_ids)

playlist_ids.extend([203,304]) #extend adds multiple values in list
print(playlist_ids)
```

```python
"""3.
Simulate removing the last played song from your playlist_ids list using
pop(), and display the removed ID along with the remaining playlist."""

playlist_ids = [101, 202, 303, 404, 505]
print(playlist_ids)

removed = playlist_ids.pop() #it removes the last vlaues of list 
print(removed) 
print(playlist_ids)
```
